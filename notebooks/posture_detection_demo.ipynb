# Posture Detection Demo

This notebook demonstrates how to use the `hunchback_detection` package to compute the posture angle and classify whether the posture is good or bad for a set of sample points. You can adapt this example to process real video frames using the command‑line scripts or by integrating the functions into your own code.

```python
from hunchback_detection.posture import calculate_angle, classify_posture

# Sample coordinates for ear, shoulder and hip in pixels (x, y)
ear = (100, 50)
shoulder = (100, 100)
hip = (100, 150)

angle = calculate_angle(ear, shoulder, hip)
status = classify_posture(angle, threshold=160)
print(f"Angle: {angle:.1f}°, Status: {status}")
```
